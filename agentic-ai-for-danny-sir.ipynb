```python
import os
from kaggle_secrets import UserSecretsClient
from smolagents import CodeAgent, DuckDuckGoSearchTool, LiteLLMModel

# 1. Load your Gemini API key from Kaggle Secrets
os.environ["GEMINI_API_KEY"] = UserSecretsClient().get_secret("GEMINI_API_KEY")

# 2. Initialize the current stable Gemini 3.8 Flash model
model = LiteLLMModel(model_id="gemini/gemini-3.8-flash") 

# 3. Create the agent and equip it with the web search tool
agent = CodeAgent(
    tools=[DuckDuckGoSearchTool()], 
    model=model
)

# 4. Give it the task
agent.run("Find the current condition of iran in regards to war and how is the populus doing?")
```
